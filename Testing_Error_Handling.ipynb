```python
# handling missing model and scaler files
from pathlib import Path
import joblib

BASE_DIR = Path.cwd()
model_path = BASE_DIR / "kmeans_model.pkl"
scaler_path = BASE_DIR / "scaler.pkl"

kmeans_model = None
scaler = None

try:
    if not model_path.exists():
        raise FileNotFoundError("K-Means model file is missing.")

    if not scaler_path.exists():
        raise FileNotFoundError("Scaler file is missing.")

    kmeans_model = joblib.load(model_path)
    scaler = joblib.load(scaler_path)

    print("Model and scaler loaded successfully.")

except FileNotFoundError as e:
    print("File error:", e)

except Exception as e:
    print(f"Loading error: {type(e).__name__}: {e}")
```

```python
### validate the model and scaler
try:
    if kmeans_model is None or scaler is None:
        raise ValueError("Model or scaler is not loaded.")

    if not hasattr(kmeans_model, "n_features_in_"):
        raise ValueError("Model is missing its expected feature information.")

    if not hasattr(scaler, "n_features_in_"):
        raise ValueError("Scaler is missing its expected feature information.")

    if kmeans_model.n_features_in_ != scaler.n_features_in_:
        raise ValueError("Model and scaler expect different numbers of features.")

    print("Model and scaler validation passed.")

except ValueError as e:
    print("Validation error:", e)
```

```python
### handle invalid customer input
import numpy as np
import pandas as pd

def validate_customer_input(customer_data, expected_features):
    try:
        customer_data = np.asarray(customer_data, dtype=float)

        if customer_data.ndim != 2:
            raise ValueError(
                "Input must be a 2D array: rows are customers, columns are features."
            )

        if customer_data.shape[1] != expected_features:
            raise ValueError(
                f"Expected {expected_features} features, "
                f"but received {customer_data.shape[1]}."
            )

        if not np.isfinite(customer_data).all():
            raise ValueError(
                "Input contains missing or infinite values."
            )

        return customer_data

    except (TypeError, ValueError) as e:
        print("Input validation error:", e)
        return None
```

```python
### test for error handling
expected_features = scaler.n_features_in_

# Test 1: Valid numeric input with the correct number of features
valid_input = np.zeros((1, expected_features))
result = validate_customer_input(valid_input, expected_features)
print("Valid input accepted:", result is not None)

# Test 2: Incorrect number of features
invalid_input = np.zeros((1, expected_features + 1))
result = validate_customer_input(invalid_input, expected_features)
print("Incorrect feature count rejected:", result is None)

# Test 3: Missing value
missing_input = np.zeros((1, expected_features))
missing_input[0, 0] = np.nan
result = validate_customer_input(missing_input, expected_features)
print("Missing value rejected:", result is None)

# Test 4: Non-numeric value
nonnumeric_input = [["invalid"] * expected_features]
result = validate_customer_input(nonnumeric_input, expected_features)
print("Non-numeric input rejected:", result is None)
```

```python
### safely haandle scaling and prediction errors
def predict_customer_cluster(customer_data):
    try:
        if kmeans_model is None or scaler is None:
            raise RuntimeError("Model or scaler is unavailable.")

        customer_data = validate_customer_input(
            customer_data,
            scaler.n_features_in_
        )

        if customer_data is None:
            return None

        scaled_data = scaler.transform(customer_data)
        prediction = kmeans_model.predict(scaled_data)

        return prediction

    except (ValueError, TypeError) as e:
        print("Input or prediction error:", e)
        return None

    except Exception as e:
        print(f"Unexpected prediction error: {type(e).__name__}: {e}")
        return None
```
